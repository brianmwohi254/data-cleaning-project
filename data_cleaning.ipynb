```python
import pandas as pd
```

```python
df = pd.read_excel("excel_data_cleaning_project_dataset.xlsx")
```

```python
df.head()
```

```python
#Find missing values
df.isna().sum()
```

```python
df.shape[0]
```

```python
df.drop_duplicates(inplace=True)
```

```python
df[~df['CustomerName'].str.istitle()]['CustomerName'] #Check for proper names that aren't capitalised
```

```python
df['CustomerName'] = df['CustomerName'].str.title() #Capitalise customer names
```

```python
df[df['Email'].str.isupper()]['Email'] #Check for emails not properly formatted
```

```python
df['Email'] = df['Email'].str.lower()
```

```python
df[~df['Product'].str.istitle()]['Product']
```

```python
df['Product'] = df['Product'].str.title()
```

```python
df['OrderDate'] = pd.to_datetime(df['OrderDate'], dayfirst=True, errors='coerce')
```

```python
df['OrderDate'] = df['OrderDate'].dt.strftime('%Y-%m-%d')
```

```python
df[~df['Quantity'] * df['UnitPrice'] == df['TotalAmount']]['TotalAmount'] #Correctly formatted
```

```python
df.to_excel("cleaned_data.xlsx", index=False)
```
